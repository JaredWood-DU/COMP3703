CAPTCHAs were designed to differentiate humans from automated systems. However, advances in deep learning and computer vision have made many traditional CAPTCHA systems vulnerable. In this assignment, you will explore this vulnerability by building a system that can recognize and decode an open-source CAPTCHA using Python, OpenCV, and a Convolutional Neural Network (CNN).

You will analyze the structure of the "Really Simple CAPTCHA" WordPress plugin and build a machine learning pipeline that can automatically identify the characters displayed in CAPTCHA images.

<b>Important Ethical Disclaimer</b> This assignment is for educational purposes only. The CAPTCHA plugin author has stated that this tool is no longer secure. You are not to use this project to attack or exploit real-world websites or systems.

---

### Learning Objectives

By completing this assignment, students will be able to:
- Understand how CAPTCHA systems generate verification images
- Apply image processing techniques to isolate individual characters
- Construct and train a Convolutional Neural Network (CNN)
- Implement contour detection using OpenCV
- Evaluate model accuracy and performance
- Analyze ethical implications of breaking security systems

---

### Background

You will be working with CAPTCHA images generated by the WordPress plugin <b>Really Simple CAPTCHA</b>, which creates 4-character codes using:
- 32 possible characters: <code>ABCDEFGHJKLMNPQRSTUVWXYZ23456789</code>
    - Characters "O" and "I" are excluded to avoid confusion
- Four random fonts per character

The plugin source code defines:
- A fixed CAPTCHA length of 4 characters
- Random font selection for each character
